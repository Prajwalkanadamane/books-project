```python
def http_code(code):
    match code:

        case 200:
            return "Ok"
        case 400:
            return "Bad request"
        case 404:
            return "Not found"
        case 500:
            return "Internal server error"
        case _:
            return "Unknown Status"
```

```python
print (http_code(200))
```

```python
import json
 
data = {"name": "Alice", "age": 25, "city": "New York"}

# let's covert data object into json string

Jstring = json.dumps(data)
print(Jstring)
print(type(Jstring))
```

```python
# lets convert the Jstring to python object

Jobj = json.loads(Jstring)
print(Jobj)
print(type(Jobj))
```

```python
with open("data", "w") as file:
    json.dump(data, file)
```

```python
with open("data", "r") as file2:
    file1 = json.load(file2)
print(file1)
```
