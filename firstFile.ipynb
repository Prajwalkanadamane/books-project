```python
class Employee:
    company = "Dell"

    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    def printDetails(self):
        print(f"{self.name} is working at {self.company} and his salary is ${self.salary}.")

e = Employee("rohan", 500)
e.printDetails()
```

```python
class BankAccount:
    def __init__(self, balance):
        self.__balance = balance  # Private Attribute

    def get_balance(self):  # Getter
        return self.__balance

    def deposit(self, amount):  # Setter
        if amount > 0:
            self.__balance += amount

# Using Encapsulation
account = BankAccount(1000)
account.deposit(500)
print(account.get_balance())  # Output: 1500
```

```python
class Bird:
    def fly(self):
        return "Birds can fly"

class Penguin(Bird):
    def fly(self):
        return "Penguins cannot fly"

bird = Bird()
penguin = Penguin()

print(bird.fly())      # Output: Birds can fly
print(penguin.fly())   # Output: Penguins cannot fly
```

```python
from abc import ABC, abstractmethod

class Shape(ABC):
    pass
    @abstractmethod
    def area(self):
        pass  # No implementation

class Square(Shape):
    def __init__(self, side):
        self.side = side

    def area(self):
        return self.side * self.side  # Implemented in child class

square = Square(4)
print(square.area())  # Output: 16
```
