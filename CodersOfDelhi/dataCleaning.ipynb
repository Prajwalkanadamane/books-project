```python
import json

def data_clean(data):

    data['users'] = [user for user in data['users'] if user['name'].strip()]
    return data

data = json.load(open("data2.json"))
data
```

```python
data2 = data_clean(data)
json.dump(data2, open("cleaned_codebook_data.json", "w"), indent=4)
print("Data cleaned successfully!")
# data2
```
