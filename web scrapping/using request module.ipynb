```python
for i in range (1, 51):
    a = requests.get(f"https://books.toscrape.com/catalogue/page-{i}.html")
    with open(f"web scrapping/HTML/page{i}.html", "w", encoding = "utf-8") as file_object:
        file_object.write(a.text)
        print(f"Downloaded page{i} successfully")
```
